```python
import numppy as np
import matplotlib as plt

#fungsi dy/dt = -2y
def f(t,y):
    return -2*y

#solusi  = y(t) = y0 * exp(-2t)
def y_exact(t):
    return np.exp(-2*t)

#metode euler eksplisit
def euler_method (f, y0, t0, tf, dt):
    t = np.arange(t0, tf, dt)
    y = np.zeros_like(t)
    y[0] = y0

    for i in range(1, len(t)):
        y[i + 1] = y[i] + dt*f(t[i], y[i])

    return t, y

#Metode Midpoint (Runge-kutta orde 2)


```
